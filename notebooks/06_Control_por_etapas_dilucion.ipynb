### 6.2 · Etapa B1: Análisis estático del costo hídrico y productivo de diluir $\text{ClO}_4^-$

## B1. ¿Cuánta agua haría falta para compensar un aumento de $\text{ClO}_4^-$? (Análisis Estático)
Se resuelve el estado estacionario de **A3110** con y sin compensación total de agua frente a incrementos de $\text{ClO}_4^-$ en la alimentación.


```python
rows = []
for f in (1.05, 1.10, 1.25, 1.50):
    R = f * (1 + R0) - 1
    Qd = R * qs0
    d = {**d0, "CP_stock": f * d0["CP_stock"]}
    xs, us, ds = pl.steady_state(alpha=0.25, d=d)
    ms = pl.metrics(xs, us, ds)
    xc, uc, dc = pl.steady_state(alpha=0.25, d=d, Q_dil=Qd)
    mc = pl.metrics(xc, uc, dc)
    rows.append({
        "ClO4 de entrada x": f,
        "dilución necesaria [m3/h]": 3600 * Qd,
        "veces la nominal": Qd / q0,
        "ppm sin comp.": ms["ppm_ClO4"],
        "Y sin comp. [%]": ms["recovery_pct"],
        "ppm con comp.": mc["ppm_ClO4"],
        "Y con comp. [%]": mc["recovery_pct"],
    })
req = pd.DataFrame(rows).set_index("ClO4 de entrada x")
req.round(2)
```

### 6.3 · Etapa B2: Cierre del lazo esclavo `FIRC 3103` frente a perturbación de válvula ($-20\%$)

## B2. Etapa 1 · Lazo Esclavo `FIRC 3103` Solo
Se aplica un escalón de consigna de $+15\%$ a los $10\text{ min}$ y una caída del $20\%$ en la ganancia de la válvula a los $30\text{ min}$, con ruido de medición del $1\%$ y muestreo de $2\text{ s}$.


```python
tun2, _ = C.autotune(pl, x0, u0, d0, dt_s=2.0)
flow_sp = lambda t: q0 * (1.15 if t >= 600 else 1.0)
dist1 = lambda t: {**d0, "valve_gain": 0.8 if t >= 1800 else 1.0}
cl = simulate(
    pl, x0, 3600, 2.0,
    C.PlantController(pl, u0, sp, tun2, enabled=("flow",), flow_sp=flow_sp, noise_rel={"flow": 0.01}, seed=1),
    disturbance=dist1, log_every=5
)
ol = simulate(
    pl, x0, 3600, 2.0,
    lambda t, x, pl_: {**u0, "Q_dil_cmd": flow_sp(t)},
    disturbance=dist1, log_every=5
)
late = lambda df: 100 * abs(df.Qd.values[df.t_h.values > 0.9].mean() / (1.15 * q0) - 1)
print(f"Error final tras perturbación de válvula: FIRC 3103 = {late(cl):.2f} % | Válvula manual = {late(ol):.2f} %")

fig, ax = plt.subplots(figsize=(7.5, 3.5))
ax.plot(cl.t_h * 60, 3600 * cl.Qd, label="Con FIRC 3103 (lazo cerrado)", color="tab:blue")
ax.plot(ol.t_h * 60, 3600 * ol.Qd, "--", label="Sin FIRC 3103 (válvula manual)", color="tab:red")
ax.plot(cl.t_h * 60, [3600 * flow_sp(t) for t in cl.t_h * 3600], "k:", label="Consigna F_303_sp")
ax.set(xlabel="Tiempo [min]", ylabel="Caudal dilución [m³/h]", title="Etapa 1: Desempeño del esclavo FIRC 3103")
ax.legend(fontsize=8)
fig.tight_layout()
fig.savefig("../docs/fig/06_etapa1_FIRC.png", dpi=150)
plt.show()
```

### 6.4 · Etapa B3: Cierre del controlador de razón `FFIC 3102` ante cambio de carga (`SSKd`)

## B3. Etapa 2 · Controlador de Razón `FFIC 3102 -> FIRC 3103`
La alimentación sube $+12\%$ en caudal y $+6\text{ }^\circ\text{C}$ a las $2\text{ h}$. Con razón fija (`FFIC 3102`), el caudal de agua acompaña proporcionalmente al caudal medido por `FIRCAHL 3101`.


```python
Qs = lambda t: qs0 * (1.12 if t >= 2 * 3600 else 1.0)
dist2 = lambda t: {**d0, "Q_stock": Qs(t), "T_feed": d0["T_feed"] + (6.0 if t >= 2 * 3600 else 0.0)}
H = 8 * 3600
fija = simulate(pl, x0, H, DT, C.PlantController(pl, u0, sp, tun, enabled=("T", "V", "flow")), disturbance=dist2, log_every=12)
razon = simulate(pl, x0, H, DT, C.PlantController(pl, u0, sp, tun, enabled=("T", "V", "ratio"), q_stock_meas=Qs, ratio_nom=R0), disturbance=dist2, log_every=12)
pd.DataFrame({
    "dilución final [m3/h]": {"consigna fija": 3600 * fija.Qd.iloc[-1], "razón fija (FFIC 3102)": 3600 * razon.Qd.iloc[-1]},
    "razón dilución/alimentación": {"consigna fija": fija.Qd.iloc[-1] / Qs(H), "razón fija (FFIC 3102)": razon.Qd.iloc[-1] / Qs(H)},
}).round(4)
```

### 6.5 · Etapa B4: Analizador `AIT 3105` y barrido de tiempo muerto ($\theta_{\text{AIT}} \in [0, 2\text{ h}]$)

## B4. Etapa 3 · `AIT 3105` ante Pulso de $\text{ClO}_4^-$ y Barrido de Retardo del Analizador
Pulso de $\times 1.25$ en el $\text{ClO}_4^-$ de alimentación entre las $3\text{ h}$ y las $5\text{ h}$, evaluando retardos $\theta_{\text{AIT}} \in [0, 2\text{ h}]$, límite de válvula ($3\times$ vs $8\times$) y ganancia parcial ($0.3$).


```python


def pulso(t):
    return d0["CP_stock"] * (1.25 if 3 * 3600 <= t < 5 * 3600 else 1.0)

dist3 = lambda t: {**d0, "CP_stock": pulso(t)}
H = 14 * 3600

def correr_ait(delay=None, cap=8.0, gain=1.0):
    if delay is None:
        ctrl = C.PlantController(pl, u0, sp, tun, enabled=("T", "V", "flow"))
    else:
        ctrl = C.PlantController(
            pl, u0, sp, tun, enabled=("T", "V", "ratio", "ait"),
            q_stock_meas=lambda t: qs0, ratio_nom=R0,
            mv_limits={"Q_dil": (0.0, cap * q0)},
            ait={"cp_true": pulso, "delay_s": delay, "gain": gain, "R_max": 3 * cap * R0}
        )
    return simulate(pl, x0, H, DT, ctrl, disturbance=dist3, log_every=12)

casos = {
    "sin AIT (dilución fija)": correr_ait(None),
    "AIT, retardo 0": correr_ait(0.0),
    "AIT, retardo 15 min": correr_ait(900.0),
    "AIT, retardo 1 h": correr_ait(3600.0),
    "AIT, retardo 2 h": correr_ait(7200.0),
    "AIT, retardo 0, válvula 3x": correr_ait(0.0, cap=3.0),
    "AIT, retardo 0, ganancia 0.3": correr_ait(0.0, gain=0.3),
}

fig, ax = plt.subplots(1, 3, figsize=(15, 4.2), sharex=True)
cols = ["tab:gray", "tab:blue", "tab:cyan", "tab:orange", "tab:red", "tab:purple", "tab:green"]
rows = []
for (name, df), col in zip(casos.items(), cols):
    ls = "--" if name.startswith("sin") else "-"
    ax[0].plot(df.t_h, df.ppm_ClO4, color=col, ls=ls, label=name)
    ax[1].plot(df.t_h, 3600 * df.Qd, color=col, ls=ls)
    ax[2].plot(df.t_h, df.recovery_pct, color=col, ls=ls)
    h = df.t_h.values
    ppm = df.ppm_ClO4.values
    Qd = 3600 * df.Qd.values
    rows.append({
        "caso": name,
        "ppm pico": ppm.max(),
        "horas sobre 40 ppm": float(np.trapezoid((ppm > 40).astype(float), h)),
        "Y mínima [%]": df.recovery_pct.min(),
        "Y final [%]": df.recovery_pct.iloc[-1],
        "dilución máx [m3/h]": Qd.max(),
        "agua extra [m3]": float(np.trapezoid(Qd - 3600 * q0, h)),
    })
ax[0].axhline(p["spec_ClO4_ppm_max"], color="r", lw=1)
ax[0].legend(fontsize=7)
for a in ax:
    a.axvspan(3, 5, color="orange", alpha=0.12)
ax[0].set(xlabel="t [h]", ylabel="ppm", title="ClO4 en producto (sombra = pulso)")
ax[1].set(xlabel="t [h]", ylabel="m³/h", title="Agua de dilución (FIRC 3103)")
ax[2].set(xlabel="t [h]", ylabel="%", title="Recuperación de KNO3")
fig.tight_layout()
fig.savefig("../docs/fig/05_AIT3105_retardo.png", dpi=150)
plt.show()

pd.DataFrame(rows).set_index("caso").round(2)
```

### 6.9 · Replay Contrafactual sobre Campaña Real SQM (10–15 Nov 2026) y Ventana de Robustez (30 días)

Carga los sensores reales de planta cada 10 min (`Análisis físicos - Seguimiento.xlsx`) y los análisis químicos cada 2 h (`Base de datos_v2.xlsx`) para evaluar qué habría ocurrido al reemplazar la dilución histórica ($\sim 20\text{ t/h}$) y la purga *on/off* por la estrategia sostenible propuesta ($0{,}5\text{ t/h}$ nominal + `AIT 3105` + purga continua `AIC 3255` + *trim* suave en $T_2$).


```python
# ==============================================================================
# 6.9 · Carga Sincronizada de Campaña Real SQM y Modelo de Replay Contrafactual
# ==============================================================================
from pathlib import Path
import unicodedata
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display


def _find_sqm_excel(keyword):
    """Busca archivos Excel en data/ o ../data/ tolerando acentos NFC/NFD de macOS."""
    def _norm(s):
        return unicodedata.normalize("NFKD", s).encode("ascii", "ignore").decode("ascii").lower()
    kw = _norm(keyword)
    for base in (Path("."), Path("..")):
        for folder in (base / "data", base):
            if folder.exists():
                for p in sorted(folder.glob("*.xlsx")):
                    if kw in _norm(p.name) and not p.name.startswith("~$"):
                        return p
    return None


def cargar_ventana_real_sqm(
    t_inicio="2026-11-10 22:00:00",
    t_fin="2026-11-15 05:00:00",
    usar_sensores_4d=True,
):
    """Sincroniza sensores de planta (10 min) y química (2 h) desde los Excel de SQM."""
    ruta_fis = _find_sqm_excel("analisis fisicos")
    ruta_v2  = _find_sqm_excel("base de datos_v2")
    if ruta_v2 is None:
        raise FileNotFoundError("Copia 'Base de datos_v2.xlsx' en la carpeta data/ del repositorio.")

    xl = pd.ExcelFile(ruta_v2)

    def leer_quim(hoja, cols_map):
        d = xl.parse(hoja)
        ts = pd.to_datetime(
            pd.to_datetime(d["Fecha"]).dt.strftime("%Y-%m-%d") + " " + d["Hora"].astype(str),
            errors="coerce",
        )
        d = d.assign(TS=ts).dropna(subset=["TS"]).set_index("TS").sort_index()
        d = d[~d.index.duplicated(keep="last")].loc[t_inicio:t_fin]
        out = pd.DataFrame(index=d.index)
        for orig, nuevo in cols_map.items():
            out[nuevo] = pd.to_numeric(d[orig], errors="coerce")
        return out

    if usar_sensores_4d and ruta_fis is not None:
        sens = pd.read_excel(ruta_fis, sheet_name="Data sensores", header=1)
        sens.columns = [str(c).strip() for c in sens.columns]
        sens["TS"] = pd.to_datetime(sens["Unnamed: 1"], errors="coerce")
        sens = sens.dropna(subset=["TS"]).set_index("TS").sort_index().loc[t_inicio:t_fin]

        df = pd.DataFrame(index=sens.index)
        df["Q_SSKd_m3_h"]  = pd.to_numeric(sens["SSKd"], errors="coerce").interpolate().bfill().ffill()
        df["Q_dil_real"]   = (pd.to_numeric(sens["Agua de dilución"], errors="coerce") / 1000.0).interpolate().bfill().ffill()
        df["Q_purga_real"] = pd.to_numeric(sens["Flujo Ml purga"], errors="coerce").clip(lower=0.0).interpolate().bfill().ffill()
        df["T_A3110_real"] = pd.to_numeric(sens["Temperatura CRY A3110"], errors="coerce").interpolate().bfill().ffill()
        df["T_A3210_real"] = pd.to_numeric(sens["Temperatura CRY A3210"], errors="coerce").interpolate().bfill().ffill()
        df["Nivel_A3110"]  = pd.to_numeric(sens["Nivel A3110"], errors="coerce").interpolate().bfill().ffill()
        df["LV3101_real"]  = pd.to_numeric(sens["Valvula LV 3101"], errors="coerce").interpolate().bfill().ffill()
    else:
        d_ssk_raw = xl.parse("SSK")
        df = pd.DataFrame({
            "TS": pd.to_datetime(d_ssk_raw["Unnamed: 13"], errors="coerce").dt.round("10min"),
            "Q_SSKd_m3_h": pd.to_numeric(d_ssk_raw["Unnamed: 15"], errors="coerce"),
            "Q_dil_real": pd.to_numeric(d_ssk_raw["Unnamed: 16"], errors="coerce") / 1000.0,
        }).dropna(subset=["TS"]).set_index("TS").sort_index().loc[t_inicio:t_fin]
        df = df[~df.index.duplicated(keep="last")]
        df["Q_SSKd_m3_h"]  = df["Q_SSKd_m3_h"].clip(lower=120.0).interpolate().bfill().ffill()
        df["Q_dil_real"]   = df["Q_dil_real"].clip(lower=0.0, upper=35.0).interpolate().bfill().ffill()
        df["Q_purga_real"] = 20.43
        df["T_A3110_real"] = 57.48
        df["T_A3210_real"] = 43.04

    ssk_q   = leer_quim("SSK", {"ClO4 IC": "ClO4_SSK", "Mg": "Mg_SSK", "K": "K_SSK"})
    mlk_q   = leer_quim("Purga-Recirculación", {"ClO4 IC": "ClO4_MLK_real", "Mg": "Mg_MLK_real"})
    ref_q   = leer_quim("Cristal Refinado", {"ClO4": "ClO4_Refi_real"})
    mlref_q = leer_quim("ML Refi", {"ClO4 IC": "ClO4_MLRefi_real"})

    df = df.join([ssk_q, mlk_q, ref_q, mlref_q], how="left")
    for col in ["ClO4_SSK", "Mg_SSK", "K_SSK", "ClO4_MLK_real", "Mg_MLK_real", "ClO4_MLRefi_real"]:
        df[col] = df[col].interpolate("time").ffill().bfill()

    df["t_h"] = (df.index - df.index[0]).total_seconds() / 3600.0
    return df


class _PIControllerDiscrete:
    """Controlador PI discreto con anti-windup por retrocálculo."""
    def __init__(self, Kc, tau_I_h, sp, u_bias, u_min, u_max):
        self.Kc, self.tau_I, self.sp = Kc, tau_I_h, sp
        self.u_bias, self.u_min, self.u_max = u_bias, u_min, u_max
        self.I = 0.0

    def step(self, pv, dt_h):
        e = self.sp - pv
        u_unsat = self.u_bias + self.Kc * e + self.I
        u_sat = float(np.clip(u_unsat, self.u_min, self.u_max))
        aw = (u_sat - u_unsat) / max(self.tau_I, 0.1)
        self.I += (self.Kc / self.tau_I * e + aw) * dt_h
        return u_sat


def simular_replay_npt3(df_in, mode="baseline_sqm", theta_ait_h=0.25, lab_sampling_h=0.0, mg_offset=0.0):
    """Simula el tren acoplado Área 2000 -> 3000 -> 6000 sobre la ventana real de SQM."""
    dt_h = 10.0 / 60.0
    n = len(df_in)

    C_MLK    = float(df_in["ClO4_MLK_real"].iloc[0])
    Mg_MLK   = float(df_in["Mg_MLK_real"].iloc[0]) + mg_offset
    C_MLRefi = float(df_in["ClO4_MLRefi_real"].iloc[0])
    T2       = float(df_in["T_A3210_real"].iloc[0])
    Q_dil    = float(df_in["Q_dil_real"].iloc[0]) if mode == "baseline_sqm" else 0.50

    a_valve = 1.0 - np.exp(-dt_h / 0.05)
    a_T2    = 1.0 - np.exp(-dt_h / 0.50)
    a_B3520 = 1.0 - np.exp(-dt_h / 2.20)
    a_refi  = 1.0 - np.exp(-dt_h / 3.50)

    d_steps = max(int(round(theta_ait_h / dt_h)), 1)
    ait_buf = [float(df_in["ClO4_SSK"].iloc[0])] * d_steps

    pid_aic = _PIControllerDiscrete(Kc=-35.0, tau_I_h=1.0, sp=4.25, u_bias=28.0, u_min=18.0, u_max=65.0)
    pid_t2  = _PIControllerDiscrete(Kc=-2.2,  tau_I_h=1.2, sp=4.30, u_bias=43.0, u_min=43.0, u_max=44.3)

    C_MLK_meas, Mg_MLK_meas = C_MLK, Mg_MLK
    rows = []

    for k in range(n):
        t_h      = float(df_in["t_h"].iloc[k])
        Q_sskd   = max(float(df_in["Q_SSKd_m3_h"].iloc[k]), 50.0)
        C_ssk    = float(df_in["ClO4_SSK"].iloc[k])
        T2_real  = float(df_in["T_A3210_real"].iloc[k])
        Q_dil_r  = float(df_in["Q_dil_real"].iloc[k])
        Q_purg_r = float(df_in["Q_purga_real"].iloc[k])

        ait_buf.append(C_ssk)
        C_ait = ait_buf.pop(0)

        if lab_sampling_h > 0:
            if k % max(int(round(lab_sampling_h / dt_h)), 1) == 0:
                C_MLK_meas, Mg_MLK_meas = C_MLK, Mg_MLK
        else:
            C_MLK_meas, Mg_MLK_meas = C_MLK, Mg_MLK

        C_sat = (4.92 + 0.14 * (T2 - 43.0)) * np.exp(-0.022 * (Mg_MLK - 30.5))

        if mode == "baseline_sqm":
            Q_dil, Q_purga, T2 = Q_dil_r, Q_purg_r, T2_real
        elif mode == "dil_05_only":
            R0 = 0.50 / 243.2
            R_cmd = R0 + 0.006 * max(C_ait - 4.40, 0.0)
            Q_dil += a_valve * (float(np.clip(Q_sskd * R_cmd, 0.20, 2.50)) - Q_dil)
            Q_purga, T2 = Q_purg_r, T2_real
        else:
            R0 = 0.50 / 243.2
            R_cmd = R0 + 0.005 * max(C_ait - 4.60, 0.0) + 0.0005 * max(Mg_MLK_meas - 31.0, 0.0)
            Q_dil += a_valve * (float(np.clip(Q_sskd * R_cmd, 0.20, 2.50)) - Q_dil)

            C_sat_est = (4.92 + 0.14 * (T2 - 43.0)) * np.exp(-0.022 * (Mg_MLK_meas - 30.5))
            pid_aic.sp = min(4.45, C_sat_est - 0.35)
            ff_purge = 12.0 * max(C_ait - 4.80, 0.0) + 2.5 * max(Mg_MLK_meas - 31.2, 0.0)
            Q_purga = float(np.clip(pid_aic.step(C_MLK_meas, dt_h) + ff_purge, 18.0, 65.0))

            pid_t2.sp = min(4.50, C_sat_est - 0.28)
            T2 += a_T2 * (pid_t2.step(C_MLK_meas, dt_h) - T2)

        dil_ratio = (Q_sskd + 20.0) / (Q_sskd + Q_dil)
        purge_factor = (20.4 + 95.0) / (Q_purga + 95.0)

        if mode == "baseline_sqm":
            C_MLK_eq  = float(df_in["ClO4_MLK_real"].iloc[k])
            Mg_MLK_eq = float(df_in["Mg_MLK_real"].iloc[k]) + mg_offset
        else:
            C_MLK_eq  = float(df_in["ClO4_MLK_real"].iloc[k]) * dil_ratio * purge_factor
            Mg_MLK_eq = (float(df_in["Mg_MLK_real"].iloc[k]) + mg_offset) * dil_ratio * (0.35 + 0.65 * purge_factor)

        C_MLK  += a_B3520 * (C_MLK_eq - C_MLK)
        Mg_MLK += a_B3520 * (Mg_MLK_eq - Mg_MLK)

        eta_ratio = max(1.0 + 0.025 * (Mg_MLK - 30.5), 0.78)
        w_moist   = 0.0286 * (eta_ratio ** 0.30)

        excess = max(C_MLK - C_sat, 0.0)
        coprecip_crudo_ppm = 2600.0 * (excess ** 1.15)
        brine_crudo_ppm    = (w_moist * C_MLK * 1000.0 / 1.40) * 1.45
        ClO4_crudo_ppm     = brine_crudo_ppm + coprecip_crudo_ppm

        C_MLRefi += a_refi * ((0.90 + 0.0028 * ClO4_crudo_ppm) - C_MLRefi)
        ppm_refi_raw = 11.0 + 0.095 * brine_crudo_ppm * (C_MLRefi / 1.50)**0.35 + 0.058 * coprecip_crudo_ppm * (C_MLRefi / 1.50)**0.45
        ppm_refi = max(20.0, ppm_refi_raw)

        dKNO3_net = 0.38 * (Q_dil_r - Q_dil) - 0.055 * (Q_purga - Q_purg_r) - 0.85 * (T2 - T2_real)

        rows.append({
            "t_h": t_h, "Q_dil_t_h": Q_dil, "Q_purga_m3_h": Q_purga, "T2_C": T2,
            "C_MLK_g_L": C_MLK, "C_sat_g_L": C_sat, "Mg_MLK_g_L": Mg_MLK,
            "ClO4_refi_ppm": ppm_refi, "dKNO3_net_t_h": dKNO3_net,
        })

    return pd.DataFrame(rows, index=df_in.index)
```
